## Vectorizing Logistic Regression
### Forward Propagation
For $m=1$ (one example),

$$
\mathbf{x} = 
\begin{pmatrix}
x_1 \\
x_2 \\
\dots \\
x_n
\end{pmatrix}
,\qquad
y\in \{0,1\}
$$

Loss function:

$$
\begin{align}
z &= \mathbf{w}^{T}\mathbf{x}+b \\
\hat{y} &= \sigma{(z)} \\
\mathcal{L}(\mathbf{w},b) &= -y\log{\hat{y}}-(1-y)\log{(1-\hat{y})}
\end{align}
$$

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

rng = np.random.default_rng()

x = rng.uniform(low=0,
                high=1,
                size=(100,1))

y = rng.integers(low=0, high=1)

w = rng.uniform(low=-10,high=10,size=(100,1))
b = rng.uniform(low=-10,high=10)
z = np.dot(w.T,x)+b
y_pred = sigmoid(z)
loss = -y*np.log(y_pred) - (1-y) * np.log(1-y_pred)

def log_loss(x,y,w,b):
    z = np.dot(w.T,x)+b
    y_pred = sigmoid(z)
    loss = - y * np.log(y_pred) - (1-y) * np.log(1-y_pred)
    return loss
```

For $m$ examples
$$
\mathbf{X} = \begin{pmatrix}
x_1^{(1)} & x_1^{(2)} & \cdots & x_1^{(i)} & \cdots & x_1^{(m)} \\
x_2^{(1)} & x_2^{(2)} & \cdots & x_2^{(i)} & \cdots & x_2^{(m)} \\
\vdots & \vdots & \ddots & \vdots & \ddots & \vdots \\
x_j^{(1)} & x_j^{(2)} & \cdots & x_j^{(i)} & \cdots & x_j^{(m)} \\
\vdots & \vdots & \ddots & \vdots & \ddots & \vdots \\
x_{n}^{(1)} & x_n^{(2)} & \cdots & x_n^{(i)} & \cdots & x_n^{(m)}
\end{pmatrix}
,\qquad
\mathbf{y} = (y^{(1)},y^{(2)},\dots,y^{(m)})
$$

Loss function:
$$
\begin{align*}
\mathbf{z} &= \mathbf{w}^{T}\mathbf{X} + b \\
\mathbf{\hat{y}} &= \sigma{(\mathbf{z})} \\
\mathbf{\mathcal{L}}(\mathbf{w},b) &= -\mathbf{y}\log{\mathbf{\hat{y}}}-(1-\mathbf{y})\log{(1-\mathbf{\hat{y}})}
\end{align*}
$$

Cost Function:
$$
J(\mathbf{w},b) = \frac{1}{m}\sum_i{\mathbf{\mathcal{L}}(\mathbf{w},b)}
$$

```python
X = rng.uniform(low=0.01,high=1,size=(100,1000)) #100 features x 1000 examples
y = rng.integers(low=0,high=1,size=(1,1000))

def log_cost(X,y,w,b):
    z = np.dot(w.T,X) + b
    y_hat = sigmoid(z)
    log_loss = - y * np.log(y_hat) - (1-y) * np.log(1-y_hat)
    log_cost = (1/X.shape[1]) * np.sum(log_loss, axis=1)
    return log_cost.item()

log_cost(X,y,w,b)
```
