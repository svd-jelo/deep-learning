### Vectorizing Logistic Regression

For $m=1$ (one example),

$$
\mathbf{x} = 
\begin{pmatrix}
x_1 \\
x_2 \\
\dots \\
x_n
\end{pmatrix}
,\qquad
y\in \{0,1\}
$$

Loss function:

$$
\begin{align}
z &= \mathbf{w}^{T}\mathbf{x}+b \\
\hat{y} &= \sigma{(z)} \\
\mathcal{L}(\mathbf{w},b) &= -y\log{\hat{y}}-(1-y)\log{(1-\hat{y})}
\end{align}
$$
